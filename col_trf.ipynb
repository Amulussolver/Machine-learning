```python
import requests
from io import StringIO
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
```

```python
url="https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day28-column-transformer/covid_toy.csv"
req=requests.get(url).text
data=StringIO(req)
df=pd.read_csv(data)
df.tail()
```

```python
from sklearn.compose import ColumnTransformer
trans=ColumnTransformer(transformers=[
    ("trf1",SimpleImputer(),[2]),("trf2",OrdinalEncoder(categories=[['Mild',"Strong"]]),[3]),("trf3",OneHotEncoder(sparse_output=False,drop="first"),[1,4])
],remainder="passthrough")
```

```python
x=df.iloc[:,:-1]
xtrf=trans.fit_transform(x)
```
