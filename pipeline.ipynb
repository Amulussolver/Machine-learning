```python
import requests
from io import StringIO
import numpy as np 
import pandas as pd
```

```python
url="https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day20-univariate-analysis/train.csv"
req=requests.get(url).text
data=StringIO(req)
df=pd.read_csv(data)
df.head()
```

```python
df.drop(columns=["PassengerId","Name","Ticket","Cabin"],inplace=True)
```

```python
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop("Survived",axis=1),df.Survived,test_size=0.32,random_state=0)
```

```python
X_train
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler
from sklearn.model_selection import cross_val_score,GridSearchCV

from sklearn.tree import DecisionTreeClassifier
from sklearn.compose import ColumnTransformer
```

```python
trf1=ColumnTransformer(transformers=[
    ("trsdf",SimpleImputer(strategy="mean"),[2]),
    ("ldjkal",SimpleImputer(strategy="most_frequent"),[6])
  
    
],remainder="passthrough")
```

```python
trf2=ColumnTransformer(transformers=[  ("ssdjk",OneHotEncoder(sparse_output=False,drop="first"),[1,6])],remainder="passthrough")
trf3=ColumnTransformer(transformers=[  ("ssdjkfg",MinMaxScaler(),slice(0,9))],remainder="passthrough")
```

```python
trf4=DecisionTreeClassifier()
```

```python
from sklearn import set_config
set_config(display="diagram")
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipe=Pipeline([("trf1",trf1),("trf2",trf2),("trf3",trf3),("trf4",trf4)])
```

```python
pipe.fit(X_train,y_train)
```
