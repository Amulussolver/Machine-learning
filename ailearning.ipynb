```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

```python
data =pd.read_csv('Housing.csv')
```

```python
data
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(data.drop(columns=["price"]),data['price'],test_size=0.34,random_state=32)
```

```python
x_train
```

```python
x_train['mainroad']=x_train["mainroad"].map({"yes":1,"no":0})
x_train['guestroom']=x_train["guestroom"].map({"yes":1,"no":0})
x_train['basement']=x_train["basement"].map({"yes":1,"no":0})
x_train['hotwaterheating']=x_train["hotwaterheating"].map({"yes":1,"no":0})
x_train['airconditioning']=x_train["airconditioning"].map({"yes":1,"no":0})
x_train['prefarea']=x_train["prefarea"].map({"yes":1,"no":0})
```

```python
x_train
```

```python
Oe=OrdinalEncoder(categories=[["furnished","semi-furnished","unfurnished"]])
x_train["furnishingstatus"]=Oe.fit_transform(x_train.iloc[:,11:12])
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression as leg
lreg=leg()
lreg.fit(x_train,y_train)
```

```python
lreg.predict([[3680	,3,	2	,2	,1,	0	,0,	0	,0	,0	,0	,1.0]])
```

```python
import pickle
with open ("house_price_model.pkl","wb") as file:
    pickle.dump(lreg,file)
with open("encoder.pkl","wb") as file:
        pickle.dump(Oe,file)
```

```python
with open ("house_price_model.pkl","rb") as file:
    my_model=pickle.load (file)
with open("encoder.pkl","rb") as file:
        encoder=pickle.load(file)
```
